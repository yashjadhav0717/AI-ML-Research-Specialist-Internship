```python
# Import Pandas for data manipulation and analysis
import pandas as pd

# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib for creating visualizations
import matplotlib.pyplot as plt

# Import Seaborn for advanced statistical visualizations
import seaborn as sns

# Import Path for handling file paths in a platform-independent way
from pathlib import Path

# Configure the default visual style for Seaborn
sns.set_theme(style="whitegrid")

# Set the default figure size for better readability
plt.rcParams["figure.figsize"] = (10, 6)

# Display all columns when viewing DataFrames
pd.set_option("display.max_columns", None)

# Display up to 100 rows when necessary
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")
```

```python
# Define the folder containing the dataset
dataset_folder = Path(
    r"C:\Users\Yash\OneDrive\Desktop\AI-ML-Research-Specialist-Internship\Task-03-Real-World Dataset Analysis Project\dataset"
)

# Define the complete path to the Global Superstore Excel file
dataset_path = dataset_folder / "Global Superstore.xls"

# Check whether the dataset file exists
if dataset_path.exists():
    print("Dataset found successfully!")
    print("Dataset path:", dataset_path)
else:
    print("Dataset not found.")
    print("Please check the dataset folder and file name.")
```

```python
# Calculate the total number of missing values in each column
missing_values = df.isnull().sum()

# Calculate the percentage of missing values in each column
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Combine both measurements into one DataFrame
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

# Sort columns by the number of missing values
missing_summary = missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)

# Display the missing-value summary
missing_summary
```

```python
# Select only columns containing missing values
missing_data = missing_summary[
    missing_summary["Missing Values"] > 0
]

# Check whether any missing values exist before plotting
if not missing_data.empty:
    
    # Create a bar chart showing missing values
    plt.figure(figsize=(12, 6))
    
    sns.barplot(
        x=missing_data["Missing Values"],
        y=missing_data.index
    )
    
    # Add chart title and labels
    plt.title("Missing Values by Column")
    plt.xlabel("Number of Missing Values")
    plt.ylabel("Column")
    
    # Display the chart
    plt.show()

else:
    print("No missing values found in the dataset.")
```

```python
# Count the number of completely duplicated rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
# Remove duplicate records from the dataset
# Keeping the first occurrence of each duplicated record
df = df.drop_duplicates()

# Display the new dataset shape after duplicate removal
print("Dataset shape after removing duplicates:", df.shape)

# Verify that duplicate records have been removed
print("Remaining duplicate records:", df.duplicated().sum())
```

```python
# Convert Order Date into Pandas datetime format
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

# Convert Ship Date into Pandas datetime format
df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)

# Display the data types after conversion
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
# Count invalid or missing values created during date conversion
invalid_order_dates = df["Order Date"].isna().sum()
invalid_ship_dates = df["Ship Date"].isna().sum()

print("Invalid Order Dates:", invalid_order_dates)
print("Invalid Ship Dates:", invalid_ship_dates)
```

```python
# Extract the year from the Order Date
df["Year"] = df["Order Date"].dt.year

# Extract the month number from the Order Date
df["Month"] = df["Order Date"].dt.month

# Extract the month name from the Order Date
df["Month_Name"] = df["Order Date"].dt.month_name()

# Extract the quarter from the Order Date
df["Quarter"] = df["Order Date"].dt.quarter

# Calculate the number of days taken to ship each order
df["Shipping_Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

# Display the newly created columns
df[
    [
        "Order Date",
        "Ship Date",
        "Year",
        "Month",
        "Month_Name",
        "Quarter",
        "Shipping_Days"
    ]
].head()
```

```python
# Identify all numerical columns in the dataset
numeric_columns = df.select_dtypes(
    include=np.number
).columns

# Display the numerical columns
print("Numerical Columns:")
print(list(numeric_columns))
```

```python
# Identify all categorical/object columns in the dataset
categorical_columns = df.select_dtypes(
    include="object"
).columns

# Display the categorical columns
print("Categorical Columns:")
print(list(categorical_columns))
```

```python
# Display the number of unique values in each column
unique_values = df.nunique().sort_values(ascending=False)

# Convert the result into a DataFrame for readability
unique_summary = unique_values.to_frame(
    name="Unique Values"
)

# Display the result
unique_summary
```

```python
# Count the number of records in each product category
category_counts = df["Category"].value_counts()

# Display the category counts
category_counts
```

```python
# Create a bar chart showing the number of records in each category
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Category"
)

# Add chart title and axis labels
plt.title("Number of Orders by Category")
plt.xlabel("Category")
plt.ylabel("Number of Orders")

# Display the chart
plt.show()
```

```python
# Calculate total sales for each product category
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Display total sales by category
category_sales
```

```python
# Create a bar chart for total sales by category
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

# Add chart title and labels
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

# Display the chart
plt.show()
```

```python
# Calculate total profit for each product category
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

# Display total profit by category
category_profit
```

```python
# Calculate total sales for each region
region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Display regional sales
region_sales
```

```python
# Create a horizontal bar chart showing sales by region
plt.figure(figsize=(10, 6))

sns.barplot(
    x=region_sales.values,
    y=region_sales.index
)

# Add chart title and labels
plt.title("Total Sales by Region")
plt.xlabel("Total Sales")
plt.ylabel("Region")

# Display the chart
plt.show()
```

```python
# Calculate total profit for each region
region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

# Display regional profit
region_profit
```

```python
# Group sales by month and calculate total monthly sales
monthly_sales = (
    df.groupby(
        df["Order Date"].dt.to_period("M")
    )["Sales"]
    .sum()
)

# Display monthly sales
monthly_sales
```

```python
# Convert the PeriodIndex into timestamps for plotting
monthly_sales_plot = monthly_sales.copy()
monthly_sales_plot.index = monthly_sales_plot.index.to_timestamp()

# Create a line chart showing monthly sales trends
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales_plot.index,
    monthly_sales_plot.values
)

# Add chart title and labels
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

# Rotate x-axis labels for readability
plt.xticks(rotation=45)

# Display the chart
plt.show()
```

```python
# Calculate total sales for every product
product_sales = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Select the top 10 products by sales
top_10_products = product_sales.head(10)

# Display the top 10 products
top_10_products
```

```python
# Create a horizontal bar chart for the top 10 products
plt.figure(figsize=(12, 7))

sns.barplot(
    x=top_10_products.values,
    y=top_10_products.index
)

# Add chart title and labels
plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")

# Display the chart
plt.show()
```

```python
# Calculate total profit for each product
product_profit = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values()
)

# Identify the 10 products with the lowest total profit
loss_making_products = product_profit.head(10)

# Display the results
loss_making_products
```

```python
# Create a scatter plot to investigate the relationship
# between discount percentage and profit
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit",
    alpha=0.5
)

# Add chart title and labels
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

# Display the chart
plt.show()
```

```python
# Select important numerical business variables for correlation analysis
correlation_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

# Calculate the Pearson correlation matrix
correlation_matrix = df[correlation_columns].corr()

# Display the correlation matrix
correlation_matrix
```

```python
# Create a heatmap to visually represent correlations
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

# Add chart title
plt.title("Correlation Matrix of Key Numerical Variables")

# Display the chart
plt.show()
```

```python
# Create a histogram to understand the distribution of sales values
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Sales",
    bins=30,
    kde=True
)

# Add chart title and labels
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

# Display the chart
plt.show()
```

```python
# Create a histogram to understand the distribution of profit values
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Profit",
    bins=30,
    kde=True
)

# Add chart title and labels
plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")

# Display the chart
plt.show()
```

```python
# Create a box plot to identify potential outliers in sales
plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["Sales"]
)

# Add chart title and label
plt.title("Sales Distribution and Potential Outliers")
plt.xlabel("Sales")

# Display the chart
plt.show()
```

```python
# Create a box plot to identify potential outliers in profit
plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["Profit"]
)

# Add chart title and label
plt.title("Profit Distribution and Potential Outliers")
plt.xlabel("Profit")

# Display the chart
plt.show()
```

```python
# Calculate total sales generated by each customer
customer_sales = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Select the top 10 customers
top_customers = customer_sales.head(10)

# Display the top customers
top_customers
```

```python
# Calculate important business metrics for each customer segment
segment_analysis = (
    df.groupby("Segment")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Average_Discount=("Discount", "mean")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

# Display segment-level analysis
segment_analysis
```

```python
# Calculate total sales across the entire dataset
total_sales = df["Sales"].sum()

# Calculate total profit across the entire dataset
total_profit = df["Profit"].sum()

# Calculate total quantity sold
total_quantity = df["Quantity"].sum()

# Calculate the average sales value per record
average_sales = df["Sales"].mean()

# Calculate the average profit per record
average_profit = df["Profit"].mean()

# Display the key performance indicators
print("========== KEY PERFORMANCE INDICATORS ==========")
print(f"Total Sales       : {total_sales:,.2f}")
print(f"Total Profit      : {total_profit:,.2f}")
print(f"Total Quantity    : {total_quantity:,.0f}")
print(f"Average Sales     : {average_sales:,.2f}")
print(f"Average Profit    : {average_profit:,.2f}")
```

```python
# Calculate the overall profit margin
# Profit Margin = Total Profit / Total Sales * 100
profit_margin = (total_profit / total_sales) * 100

# Display the profit margin
print(f"Overall Profit Margin: {profit_margin:.2f}%")
```
