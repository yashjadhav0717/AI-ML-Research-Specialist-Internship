```python
# ============================================================
# CELL 2 — IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from IPython.display import display

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

# Visualization style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully.")
```

```python
# ============================================================
# CELL 3 — PROJECT PATHS
# ============================================================

project_folder = Path(
    r"C:\Users\Yash\OneDrive\Desktop\AI-ML-Research-Specialist-Internship\Task-05-Analytics Report & Insights"
)

dataset_folder = project_folder / "dataset"
visualization_folder = project_folder / "visualizations"
report_folder = project_folder / "report"

# Create required folders if they don't exist
dataset_folder.mkdir(parents=True, exist_ok=True)
visualization_folder.mkdir(parents=True, exist_ok=True)
report_folder.mkdir(parents=True, exist_ok=True)

# Expected dataset path
dataset_path = dataset_folder / "cleaned_global_superstore.csv"

print("Project folder:")
print(project_folder)

print("\nDataset path:")
print(dataset_path)

print("\nDataset exists:", dataset_path.exists())
```

```python
# ============================================================
# CELL 4 — VERIFY DATASET
# ============================================================

if not dataset_path.exists():
    print("ERROR: Dataset file was not found.")
    print("\nPlease make sure this file exists:")
    print(dataset_path)
else:
    print("Dataset found successfully!")
    print("File:", dataset_path.name)
```

```python
# ============================================================
# CELL 5 — LOAD DATASET
# ============================================================

df = pd.read_csv(dataset_path)

print("Dataset loaded successfully.")
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
```

```python
# ============================================================
# CELL 6 — DATASET PREVIEW
# ============================================================

display(df.head())
```

```python
# ============================================================
# CELL 7 — DATASET INFORMATION
# ============================================================

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

print("\nDataset Information:")
df.info()
```

```python
# ============================================================
# CELL 8 — DATA QUALITY CHECK
# ============================================================

# Missing values
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
}).sort_values(
    by="Missing Values",
    ascending=False
)

print("Missing Value Analysis:")
display(missing_summary)

# Duplicate records
duplicate_count = df.duplicated().sum()

print(f"\nDuplicate records: {duplicate_count:,}")
```

```python
# ============================================================
# CELL 9 — DATA PREPARATION
# ============================================================

# Convert date columns
if "Order Date" in df.columns:
    df["Order Date"] = pd.to_datetime(
        df["Order Date"],
        errors="coerce"
    )

if "Ship Date" in df.columns:
    df["Ship Date"] = pd.to_datetime(
        df["Ship Date"],
        errors="coerce"
    )

# Create time-based columns
if "Order Date" in df.columns:

    df["Year"] = df["Order Date"].dt.year
    df["Month"] = df["Order Date"].dt.month
    df["Month_Name"] = df["Order Date"].dt.month_name()
    df["Quarter"] = df["Order Date"].dt.quarter

# Calculate shipping days if required
if "Ship Date" in df.columns and "Order Date" in df.columns:

    df["Shipping_Days"] = (
        df["Ship Date"] - df["Order Date"]
    ).dt.days

print("Data preparation completed.")

print("\nDate range:")
print(df["Order Date"].min(), "to", df["Order Date"].max())
```

```python
# ============================================================
# CELL 12 — KEY PERFORMANCE INDICATORS
# ============================================================

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()

average_sales = df["Sales"].mean()
average_profit = df["Profit"].mean()

average_discount = df["Discount"].mean()

profit_margin = (
    total_profit / total_sales * 100
)

# Optional metrics
if "Customer ID" in df.columns:
    total_customers = df["Customer ID"].nunique()
else:
    total_customers = df["Customer Name"].nunique()

if "Order ID" in df.columns:
    total_orders = df["Order ID"].nunique()
else:
    total_orders = len(df)

if "Shipping_Days" in df.columns:
    average_shipping_days = df["Shipping_Days"].mean()
else:
    average_shipping_days = np.nan

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Profit",
        "Total Quantity",
        "Average Sales",
        "Average Profit",
        "Average Discount",
        "Profit Margin",
        "Total Customers",
        "Total Orders",
        "Average Shipping Days"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_quantity,
        average_sales,
        average_profit,
        average_discount,
        profit_margin,
        total_customers,
        total_orders,
        average_shipping_days
    ]
})

display(kpi_summary)
```

```python
# ============================================================
# CELL 13 — KPI SUMMARY
# ============================================================

print("=" * 60)
print("GLOBAL SUPERSTORE — KEY PERFORMANCE INDICATORS")
print("=" * 60)

print(f"Total Sales           : ${total_sales:,.2f}")
print(f"Total Profit          : ${total_profit:,.2f}")
print(f"Total Quantity        : {total_quantity:,.0f}")
print(f"Average Sales         : ${average_sales:,.2f}")
print(f"Average Profit        : ${average_profit:,.2f}")
print(f"Average Discount      : {average_discount:.2%}")
print(f"Profit Margin         : {profit_margin:.2f}%")
print(f"Total Customers       : {total_customers:,}")
print(f"Total Orders          : {total_orders:,}")
print(f"Average Shipping Days : {average_shipping_days:.2f}")
```

```python
# ============================================================
# CELL 14 — CATEGORY ANALYSIS
# ============================================================

category_analysis = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Average_Discount=("Discount", "mean")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

category_analysis["Profit_Margin"] = (
    category_analysis["Total_Profit"]
    / category_analysis["Total_Sales"]
    * 100
)

display(category_analysis)
```

```python
# ============================================================
# CELL 15 — CATEGORY SALES VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_analysis.reset_index(),
    x="Category",
    y="Total_Sales"
)

plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.tight_layout()

plt.savefig(
    visualization_folder / "sales_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 16 — REGIONAL ANALYSIS
# ============================================================

regional_analysis = (
    df.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Average_Discount=("Discount", "mean")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

regional_analysis["Profit_Margin"] = (
    regional_analysis["Total_Profit"]
    / regional_analysis["Total_Sales"]
    * 100
)

display(regional_analysis)
```

```python
# ============================================================
# CELL 17 — REGIONAL SALES VISUALIZATION
# ============================================================

plt.figure(figsize=(11, 6))

sns.barplot(
    data=regional_analysis.reset_index(),
    x="Total_Sales",
    y="Region"
)

plt.title("Total Sales by Region")
plt.xlabel("Sales")
plt.ylabel("Region")

plt.tight_layout()

plt.savefig(
    visualization_folder / "sales_by_region.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 18 — CUSTOMER SEGMENT ANALYSIS
# ============================================================

segment_analysis = (
    df.groupby("Segment")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Average_Discount=("Discount", "mean")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

segment_analysis["Profit_Margin"] = (
    segment_analysis["Total_Profit"]
    / segment_analysis["Total_Sales"]
    * 100
)

display(segment_analysis)
```

```python
# ============================================================
# CELL 19 — SEGMENT VISUALIZATION
# ============================================================

plt.figure(figsize=(9, 6))

sns.barplot(
    data=segment_analysis.reset_index(),
    x="Segment",
    y="Total_Sales"
)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Sales")

plt.tight_layout()

plt.savefig(
    visualization_folder / "sales_by_segment.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 20 — PRODUCT ANALYSIS
# ============================================================

product_analysis = (
    df.groupby("Product Name")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

top_10_products = product_analysis.head(10)

print("Top 10 Products by Sales:")
display(top_10_products)
```

```python
# ============================================================
# CELL 21 — TOP PRODUCTS VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 7))

top_products_plot = top_10_products.sort_values(
    "Total_Sales"
)

sns.barplot(
    data=top_products_plot.reset_index(),
    x="Total_Sales",
    y="Product Name"
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")

plt.tight_layout()

plt.savefig(
    visualization_folder / "top_10_products.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 22 — LOWEST PROFIT PRODUCTS
# ============================================================

lowest_profit_products = (
    product_analysis
    .sort_values("Total_Profit")
    .head(10)
)

print("10 Products with Lowest Profit:")
display(lowest_profit_products)
```

```python
# ============================================================
# CELL 23 — MONTHLY SALES ANALYSIS
# ============================================================

monthly_sales = (
    df.groupby(
        df["Order Date"].dt.to_period("M")
    )["Sales"]
    .sum()
)

monthly_sales_plot = monthly_sales.copy()

monthly_sales_plot.index = (
    monthly_sales_plot.index.to_timestamp()
)

display(monthly_sales)
```

```python
# ============================================================
# CELL 24 — MONTHLY SALES TREND
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales_plot.index,
    monthly_sales_plot.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    visualization_folder / "monthly_sales_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 25 — DISCOUNT VS PROFIT
# ============================================================

discount_profit_corr = (
    df["Discount"].corr(df["Profit"])
)

print(
    f"Correlation between Discount and Profit: "
    f"{discount_profit_corr:.3f}"
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit",
    alpha=0.4
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.tight_layout()

plt.savefig(
    visualization_folder / "discount_vs_profit.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 26 — SALES VS PROFIT
# ============================================================

sales_profit_corr = (
    df["Sales"].corr(df["Profit"])
)

print(
    f"Correlation between Sales and Profit: "
    f"{sales_profit_corr:.3f}"
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    alpha=0.4
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.tight_layout()

plt.savefig(
    visualization_folder / "sales_vs_profit.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 27 — CORRELATION ANALYSIS
# ============================================================

correlation_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

correlation_matrix = (
    df[correlation_columns]
    .corr()
)

display(correlation_matrix)
```

```python
# ============================================================
# CELL 28 — CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Key Business Variables")

plt.tight_layout()

plt.savefig(
    visualization_folder / "correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 29 — SALES DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Sales",
    bins=30,
    kde=True
)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    visualization_folder / "sales_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 30 — PROFIT DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Profit",
    bins=30,
    kde=True
)

plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    visualization_folder / "profit_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# CELL 31 — CUSTOMER ANALYSIS
# ============================================================

customer_analysis = (
    df.groupby("Customer Name")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

top_customers = customer_analysis.head(10)

print("Top 10 Customers by Sales:")
display(top_customers)
```

```python
# ============================================================
# CELL 32 — AUTOMATIC BUSINESS FINDINGS
# ============================================================

best_category = category_analysis.index[0]
best_category_sales = category_analysis.iloc[0]["Total_Sales"]

best_region = regional_analysis.index[0]
best_region_sales = regional_analysis.iloc[0]["Total_Sales"]

best_segment = segment_analysis.index[0]
best_segment_sales = segment_analysis.iloc[0]["Total_Sales"]

best_product = top_10_products.index[0]
best_product_sales = top_10_products.iloc[0]["Total_Sales"]

lowest_profit_product = lowest_profit_products.index[0]
lowest_product_profit = lowest_profit_products.iloc[0]["Total_Profit"]

print("=" * 70)
print("KEY BUSINESS FINDINGS")
print("=" * 70)

print(
    f"\n1. Highest-performing category: "
    f"{best_category} "
    f"(${best_category_sales:,.2f} sales)"
)

print(
    f"\n2. Highest-performing region: "
    f"{best_region} "
    f"(${best_region_sales:,.2f} sales)"
)

print(
    f"\n3. Highest-performing customer segment: "
    f"{best_segment} "
    f"(${best_segment_sales:,.2f} sales)"
)

print(
    f"\n4. Top product by sales: "
    f"{best_product} "
    f"(${best_product_sales:,.2f})"
)

print(
    f"\n5. Lowest-profit product identified: "
    f"{lowest_profit_product} "
    f"(${lowest_product_profit:,.2f} profit)"
)

print(
    f"\n6. Discount-Profit correlation: "
    f"{discount_profit_corr:.3f}"
)

print(
    f"\n7. Sales-Profit correlation: "
    f"{sales_profit_corr:.3f}"
)
```
