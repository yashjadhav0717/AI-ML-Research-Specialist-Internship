```python
#NumPy Numerical Operations
math_scores = np.array(df["Math_Score"])

print("Mean Math Score:", np.mean(math_scores))
print("Maximum Math Score:", np.max(math_scores))
print("Minimum Math Score:", np.min(math_scores))
print("Standard Deviation:", np.std(math_scores))
```

```python
#NumPy Total Scores
python_scores = np.array(df["Python_Score"])

total_scores = math_scores + python_scores

print("Total Scores:")
print(total_scores)
```

```python
#NumPy Average Scores
average_scores = (
    math_scores + python_scores
) / 2

print("Average Scores:")
print(average_scores)
```

```python
#Create Average Score Column
df["Average_Score"] = (
    df["Math_Score"] + df["Python_Score"]
) / 2

df.head()
```

```python
#Create Grade Column
def calculate_grade(score):
    if score >= 90:
        return "A+"
    elif score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "D"


df["Grade"] = df["Average_Score"].apply(calculate_grade)

df[
    [
        "Name",
        "Average_Score",
        "Grade"
    ]
]
```

```python
#Grouping by Department
department_analysis = df.groupby(
    "Department"
)[
    [
        "Study_Hours",
        "Attendance",
        "Math_Score",
        "Python_Score",
        "Average_Score"
    ]
].mean()

department_analysis
```

```python
#Grouping by Gender
gender_analysis = df.groupby(
    "Gender"
)["Average_Score"].mean()

gender_analysis
```

```python
#Aggregation
department_summary = df.groupby(
    "Department"
)["Average_Score"].agg(
    [
        "mean",
        "min",
        "max",
        "count"
    ]
)

department_summary
```

```python
#Create Dirty Dataset
dirty_df = df.copy()

# Introduce missing values
dirty_df.loc[2, "Attendance"] = np.nan
dirty_df.loc[7, "Math_Score"] = np.nan

# Introduce duplicate record
dirty_df = pd.concat(
    [
        dirty_df,
        dirty_df.iloc[[0]]
    ],
    ignore_index=True
)

print("Dirty Dataset Shape:", dirty_df.shape)
dirty_df.head()
```

```python
#Detect Missing Values
missing_values = dirty_df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
#Total Missing Values
total_missing = dirty_df.isnull().sum().sum()

print("Total Missing Values:", total_missing)
```

```python
#Handle Missing Values
clean_df = dirty_df.copy()

numeric_columns = [
    "Age",
    "Study_Hours",
    "Attendance",
    "Math_Score",
    "Python_Score"
]

for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(
        clean_df[column].median()
    )

print("Missing values after treatment:")
print(clean_df.isnull().sum())
```

```python
#Detect Duplicate Records
duplicate_count = clean_df.duplicated().sum()

print("Duplicate Rows:", duplicate_count)
```

```python
#Remove Duplicate Records
clean_df = clean_df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:")
print(clean_df.shape)
```

```python
#Check Data Types
print("Data Types:")
print(clean_df.dtypes)
```

```python
#Descriptive Statistics After Cleaning
numeric_summary = clean_df[
    [
        "Age",
        "Study_Hours",
        "Attendance",
        "Math_Score",
        "Python_Score",
        "Average_Score"
    ]
].describe()

numeric_summary
```

```python
#Bar Chart
plt.figure(figsize=(10, 6))

department_avg = clean_df.groupby(
    "Department"
)["Average_Score"].mean().sort_values()

department_avg.plot(kind="bar")

plt.title("Average Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Score")
plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "bar_chart.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Bar chart saved successfully.")
```
