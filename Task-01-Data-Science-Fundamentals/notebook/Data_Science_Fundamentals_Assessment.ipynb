```python
marks = 75

if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"

print("Marks:", marks)
print("Grade:", grade)
```

## 4. Statistical Foundations

This section covers descriptive statistics including mean, median, mode, variance, standard deviation, percentiles, probability, and correlation.

```python
import statistics
import numpy as np
import pandas as pd
```

```python
#Mean, Median, Mode

data = [10, 20, 20, 30, 40, 50]

mean = statistics.mean(data)
median = statistics.median(data)
mode = statistics.mode(data)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
#Variance and Standard Deviation

variance = statistics.variance(data)
std_dev = statistics.stdev(data)

print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

```python
#Percentiles

print("25th Percentile:", np.percentile(data, 25))
print("50th Percentile:", np.percentile(data, 50))
print("75th Percentile:", np.percentile(data, 75))
```

### Probability

```python
favorable_outcomes = 1
total_outcomes = 2

probability = favorable_outcomes / total_outcomes

print("Probability of getting heads:", probability)
```

## 5. Student Dataset

A sample student dataset is created to demonstrate data manipulation, statistical analysis, exploratory data analysis, and visualization.

```python
data = {
    "Student_ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Name": [
        "Amit", "Priya", "Rahul", "Sneha",
        "Karan", "Neha", "Rohit", "Pooja"
    ],
    "Age": [20, 21, 20, 22, 21, 20, 23, 21],
    "Study_Hours": [2, 4, 3, 5, 6, 3, 7, 4],
    "Marks": [55, 72, 65, 80, 88, 68, 92, 75],
    "City": [
        "Pune", "Mumbai", "Satara", "Pune",
        "Sangli", "Satara", "Mumbai", "Pune"
    ]
}

df = pd.DataFrame(data)

df
```

```python
print("Shape:", df.shape)
```

```python
df.info()
```

```python
df.describe()
```

## 6. Data Cleaning

```python
df.isnull().sum()
```

```python
#filtering

high_scorers = df[df["Marks"] >= 80]

high_scorers
```

```python
#Sorting

sorted_df = df.sort_values("Marks", ascending=False)

sorted_df
```

```python
#Grouping

city_average = df.groupby("City")["Marks"].mean()

city_average
```

## 7. Statistical Analysis

```python
mean_marks = df["Marks"].mean()
median_marks = df["Marks"].median()
mode_marks = df["Marks"].mode()[0]
variance_marks = df["Marks"].var()
std_marks = df["Marks"].std()

print("Mean:", mean_marks)
print("Median:", median_marks)
print("Mode:", mode_marks)
print("Variance:", variance_marks)
print("Standard Deviation:", std_marks)
```

```python
print("25th Percentile:", df["Marks"].quantile(0.25))
print("50th Percentile:", df["Marks"].quantile(0.50))
print("75th Percentile:", df["Marks"].quantile(0.75))
```

## 8. Correlation Analysis

The relationship between Study Hours and Marks is analyzed using Pearson correlation.

```python
correlation = df["Study_Hours"].corr(df["Marks"])

print("Correlation:", correlation)
```

```python
#Performance Category

def performance_category(marks):
    if marks >= 85:
        return "Excellent"
    elif marks >= 70:
        return "Good"
    elif marks >= 50:
        return "Average"
    else:
        return "Poor"

df["Performance"] = df["Marks"].apply(performance_category)

df
```
